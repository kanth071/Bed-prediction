# 🔍 Exploratory Data Analysis (EDA)
Analyzing hospital patient flow, occupancy patterns, and disease trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/final_hospital_forecasting_dataset.csv")
df['Date'] = pd.to_datetime(df['Date'])
```

## 1. Admissions Over Time

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Date', y='Admissions', hue='service')
plt.title("Daily Admissions by Department")
plt.show()
```

## 2. Occupancy Distribution

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='service', y='Bed_Occupancy')
plt.title("Bed Occupancy Distribution per Sector")
plt.show()
```

## 3. Correlation Heatmap

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```
