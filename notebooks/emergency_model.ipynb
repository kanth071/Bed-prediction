# 🚑 Emergency Demand Forecasting - Meta Prophet
This notebook uses the Prophet forecasting model to predict future emergency department load, accounting for seasonality and trends.

```python
from prophet import Prophet
import pandas as pd
import matplotlib.pyplot as plt
from prophet.serialize import model_to_json

# Load data
df = pd.read_csv("../data/processed/final_hospital_forecasting_dataset.csv")
df_prophet = df[['Date', 'Emergency_Demand']].copy()
df_prophet.columns = ['ds', 'y']
df_prophet['ds'] = pd.to_datetime(df_prophet['ds'])
```

## 1. Model Fitting

```python
model = Prophet(yearly_seasonality=True, weekly_seasonality=True)
model.fit(df_prophet)
```

## 2. Future Predictions

```python
future = model.make_future_dataframe(periods=30)
forecast = model.predict(future)

model.plot(forecast)
plt.title("Emergency Demand Forecast")
plt.show()
```

## 3. Components Analysis

```python
model.plot_components(forecast)
plt.show()
```

## 4. Saving Model

```python
with open("../models/emergency_prophet.json", 'w') as f:
    f.write(model_to_json(model))
```
